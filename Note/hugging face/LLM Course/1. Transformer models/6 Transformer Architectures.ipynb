## Encoder models

encoder models指的是只使用transformer model的encoder的model.在其中的attention层中可以访问所有的初始token.这些模型经常被称为具有双向attention

这些模型的pretrain通常围绕着：通过一些方法‘污染’原句，再让模型预测出完整的句子

encoder models适合那些需要对整个句子有很好理解的任务。例如：sentence classification，named entity recognition.

代表模型有：
- BERT

## Decoder models

Decoder models的attention layer只能看到当前位置之前的单词。它的预训练主要围绕着预测下一个单词为主。

Decoder models适合需要texts generation的任务

代表模型有：
- Hugging Face SmolLM Series
- Meta’s Llama Series
- Google’s Gemma Series
- DeepSeek’s V3

### Modern Large Language Models(LLMs)

几乎所有的现代LLM使用decoder-only architecture. 

现代LLM典型的训练有两个阶段

- 预训练：预测下一个单词
- 指令微调：让models模仿指令生成输出

## Sequence-to-sequence models

编码器-解码器模型（也称为序列到序列模型）使用Transformer体系结构的两个部分。在每个阶段，编码器的注意层可以访问初始句子中的所有单词，而解码器的注意层只能访问输入中位于给定单词之前的单词。

这些模型的预训练可以采取不同的形式，但它通常涉及重建输入已被某种程度上损坏的句子（例如通过屏蔽随机单词）。T5模型的预训练包括用单个掩码特殊标记替换文本的随机范围（可以包含几个单词），然后任务是预测该掩码标记所替换的文本。

序列到序列模型最适合于根据给定输入生成新句子的任务，例如摘要、翻译或生成式问答。

代表模型
- BART
- mBART
- Marian
- T5

## Choosing the right architecture

When in doubt about which model to use, consider:

- What kind of understanding does your task need? (Bidirectional or unidirectional)
- Are you generating new text or analyzing existing text?
- Do you need to transform one sequence into another?

The answers to these questions will guide you toward the right architecture.
